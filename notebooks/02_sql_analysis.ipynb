```python
import sqlite3
import pandas as pd
```

```python
# Database Connection

conn = sqlite3.connect("../data/soccer.db")

print("Database connected")
```

```python
matches = pd.read_csv("../data/cleaned_matches.csv")

matches.head()
```

```python
matches.to_sql(
    "matches",
    conn,
    if_exists="replace",
    index=False
)

query = """
SELECT *
FROM matches
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT home_team_name, away_team_name, home_score, away_score
FROM matches
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT match_date, home_team_name, away_team_name, home_score, away_score
FROM matches
WHERE home_team_name = 'Leicester City' AND result = 'Home Win'
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT match_date, home_team_name, away_team_name, home_score, away_score, home_score + away_score AS total_goals
FROM matches
ORDER BY total_goals DESC
LIMIT 10
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT home_team_name, SUM(home_score) AS total_home_goals
FROM matches
GROUP BY home_team_name
ORDER BY total_home_goals DESC
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT home_team_name, COUNT(*) AS total_home_wins
FROM matches
WHERE result = 'Home Win'
GROUP BY home_team_name
ORDER BY total_home_wins DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT home_team_name, SUM(home_score) AS total_home_goals
FROM matches
GROUP BY home_team_name
HAVING total_home_goals > 30
ORDER BY total_home_goals DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT home_team_name, away_team_name, result, 
CASE
    WHEN result = 'Away Win' THEN 3
    WHEN result = 'Draw' THEN 1
    ELSE 0
END AS away_points
FROM matches
LIMIT 10
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT away_team_name, SUM(
CASE
    WHEN result = 'Away Win' THEN 3
    WHEN result = 'Draw' THEN 1
    ELSE 0
END) AS total_away_points
FROM matches
GROUP BY away_team_name
ORDER BY total_away_points DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT team, SUM(points) AS total_points
FROM (
    SELECT away_team_name AS team, 
    CASE
        WHEN result = 'Away Win' THEN 3
        WHEN result = 'Draw' THEN 1
        ELSE 0
    END AS points
    FROM matches

    UNION ALL

    SELECT home_team_name AS team, 
    CASE
        WHEN result = 'Home Win' THEN 3
        WHEN result = 'Draw' THEN 1
        ELSE 0
    END AS points
    FROM matches
)
GROUP BY team
ORDER BY total_points DESC

"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT result, COUNT(*) AS matches
FROM matches
GROUP BY result
ORDER BY matches DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT result, COUNT(*) / 380.0 * 100 AS percentage
FROM matches
GROUP BY result
ORDER BY percentage DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH away_points AS (
SELECT away_team_name AS team, SUM(
CASE
    WHEN result = 'Away Win' THEN 3
    WHEN result = 'Draw' THEN 1
    ELSE 0
END) AS away_points
FROM matches
GROUP BY away_team_name
),

team_points AS (
SELECT away_team_name AS team, 
    CASE
        WHEN result = 'Away Win' THEN 3
        WHEN result = 'Draw' THEN 1
        ELSE 0
    END AS points
    FROM matches

    UNION ALL

    SELECT home_team_name AS team, 
    CASE
        WHEN result = 'Home Win' THEN 3
        WHEN result = 'Draw' THEN 1
        ELSE 0
    END AS points
    FROM matches
),

total_points AS (
SELECT team, SUM(points) AS total_points
FROM team_points
GROUP BY team
)

SELECT away_points.team, away_points, total_points, away_points * 1.0 / total_points * 100 AS away_percentage
FROM away_points
JOIN total_points ON away_points.team = total_points.team
GROUP BY away_points.team
ORDER BY away_percentage DESC
LIMIT 5
"""

pd.read_sql_query(query, conn)
```

```python

import numpy as np

query = """
WITH home_points AS (
SELECT home_team_name AS team, SUM(
CASE
    WHEN result = 'Home Win' THEN 3
    WHEN result = 'Draw' THEN 1
    ELSE 0
END) AS home_points
FROM matches
GROUP BY home_team_name
),

away_points AS (
SELECT away_team_name AS team, SUM(
CASE
    WHEN result = 'Away Win' THEN 3
    WHEN result = 'Draw' THEN 1
    ELSE 0
END) AS away_points
FROM matches
GROUP BY away_team_name
)

SELECT home_points.team, home_points, away_points, home_points - away_points AS difference
FROM home_points
JOIN away_points ON home_points.team = away_points.team
GROUP BY home_points.team
ORDER BY difference DESC

"""

home_away = pd.read_sql_query(query, conn)

home_away.head()

x = np.arange(len(home_away))
width = 0.35

plt.figure(figsize=(16, 6))

plt.bar(
    x - width/2, 
    home_away["home_points"], 
    width, label="Home"
    )

plt.bar(
    x + width/2,
    home_away["away_points"], 
    width, label="Away"
    )

plt.xticks(x, home_away.team, rotation=45, ha="right")
plt.title("Home Points VS Away Points")
plt.legend()

plt.show()
```

```python
query = """
WITH team_goals AS (
SELECT home_team_name AS team, home_score AS goals
FROM matches

UNION ALL

SELECT away_team_name AS team, away_score AS goals
FROM matches
), 

total_goals AS (
SELECT team, SUM(goals) AS total_goals
FROM team_goals
GROUP BY team
),

team_conceded AS (
SELECT home_team_name AS team, away_score AS goals_conceded
FROM matches

UNION ALL

SELECT away_team_name AS team, home_score AS goals_conceded
FROM matches
),

total_conceded AS (
SELECT team, SUM(goals_conceded) AS total_conceded
FROM team_conceded
GROUP BY team
)


SELECT total_goals.team, total_goals, total_conceded, total_goals - total_conceded AS goal_difference 
FROM total_goals
JOIN total_conceded ON total_goals.team = total_conceded.team
GROUP BY total_goals.team
ORDER BY goal_difference DESC
"""

goals = pd.read_sql_query(query, conn)

x = np.arange(len(goals))
width = 0.35

plt.figure(figsize=(16, 6))

plt.bar(
    x - width/2, 
    goals["total_goals"], 
    width, label="Goals"
    )

plt.bar(
    x + width/2,
    goals["total_conceded"], 
    width, label="Goals Conceded"
    )

plt.xticks(x, goals.team, rotation=45, ha="right")
plt.title("Goals VS. Goals Conceded")
plt.legend()

plt.show()
```

```python
query = """
WITH total_points AS (
SELECT home_team_name AS team, home_points AS points
FROM matches

UNION ALL

SELECT away_team_name AS team, away_points AS points
FROM matches
)

SELECT team, SUM(points) AS points, COUNT(points) AS matches_played, ROUND(SUM(points) * 1.0 /COUNT(points), 2) points_per_match
FROM total_points
GROUP BY team
ORDER BY points_per_match DESC
"""

pd.read_sql_query(query, conn)
```
